# PS- 4B: CBOW Twitter Word Embeddings

This notebook follows the supplied CBOW example style.


```python
# Import all packages first
import pandas as pd
import numpy as np
import re
from gensim.models import Word2Vec
```

```python
# 1. Load updated tweets and clean text.
df = pd.read_csv("twitter_dataset.csv").dropna(subset=["Text"])
tweets = [re.sub(r"[^a-z\s]", "", tweet.lower()).split() for tweet in df["Text"]]
print(tweets[0])
```

```python
# 2. Train Word2Vec CBOW. sg=0 means CBOW.
cbow = Word2Vec(sentences=tweets, vector_size=100, window=5,
                min_count=2, sg=0, epochs=10)
print("Vocabulary size:", len(cbow.wv))
```

```python
# 3. Check word embeddings and similarity.
word = cbow.wv.index_to_key[0]
print("Word:", word)
print("Embedding:", cbow.wv[word][:10])
print("Similar words:", cbow.wv.most_similar(word, topn=5))
word1, word2 = cbow.wv.index_to_key[0], cbow.wv.index_to_key[1]
print("Similarity:", cbow.wv.similarity(word1, word2))
```
