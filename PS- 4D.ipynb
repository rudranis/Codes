# PS- 4D: CBOW Tweet Recommendation

This notebook follows the supplied CBOW example style.


```python
# Import all packages first
import pandas as pd
import numpy as np
import re
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# 1. Load updated tweets and clean text.
df = pd.read_csv("twitter_dataset.csv").dropna(subset=["Text"])
tweets = [re.sub(r"[^a-z\s]", "", tweet.lower()).split() for tweet in df["Text"]]
```

```python
# 2. Train Word2Vec CBOW.
cbow = Word2Vec(sentences=tweets, vector_size=100, window=5,
                min_count=2, sg=0, epochs=10)
```

```python
# 3. Convert every tweet into one average embedding.
tweet_embeddings = np.array([
    cbow.wv.get_mean_vector([word for word in tweet if word in cbow.wv])
    for tweet in tweets
])
```

```python
# 4. Recommend 5 tweets similar to user input.
user_input = input("Enter a topic or keywords: ")
query_words = [word for word in user_input.lower().split() if word in cbow.wv]
query_embedding = cbow.wv.get_mean_vector(query_words)
similarity = cosine_similarity([query_embedding], tweet_embeddings)[0]
top_indices = np.argsort(similarity)[::-1][:5]
for index in top_indices:
    print(df["Text"].iloc[index])
    print("Similarity:", round(similarity[index], 3))
    print("-" * 50)
```
