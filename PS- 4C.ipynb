# PS- 4C: TensorFlow CBOW for Document 3

This notebook follows the supplied CBOW example style.


```python
# Import all packages first
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Dense, Lambda
```

```python
# 1. Read updated document3.txt and convert words to numbers.
with open("document3.txt", "r", encoding="utf-8") as file:
    text = file.read().lower()
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])
words = tokenizer.texts_to_sequences([text])[0]
vocab_size = len(tokenizer.word_index) + 1
print("Total words:", len(words))
print("Vocabulary size:", vocab_size)
```

```python
# 2. Create CBOW context words and target words.
window = 2
context, target = [], []
for i in range(window, len(words) - window):
    context.append(words[i-window:i] + words[i+1:i+window+1])
    target.append(words[i])
context = np.array(context)
target = to_categorical(target, vocab_size)
print("Context shape:", context.shape)
```

```python
# 3. Build the TensorFlow CBOW model.
model = Sequential([
    Embedding(vocab_size, 100, input_length=4),
    Lambda(lambda x: tf.reduce_mean(x, axis=1)),
    Dense(vocab_size, activation="softmax")
])
model.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])
model.fit(context, target, epochs=100, verbose=1)
```

```python
# 4. Show actual and predicted words.
predictions = model.predict(context[:5])
for i in range(5):
    actual = tokenizer.index_word[np.argmax(target[i])]
    predicted = tokenizer.index_word[np.argmax(predictions[i])]
    print("Actual:", actual, "| Predicted:", predicted)
```
